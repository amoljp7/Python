7. Write a python program to find the longest words.

```python
f = open('sample.txt', 'r')

words = f.read().split()

max_len = 0
longest_words = []

for word in words:
    if len(word) > max_len:
        max_len = len(word)
        longest_words = [word]
    elif len(word) == max_len:
        longest_words.append(word)

print("Longest word = ", longest_words)

f.close()
```

9. Write a Python program to count the frequency of words in a file.


```python
f = open('sample.txt', 'r')

text = f.read()
words = text.split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)

f.close()
```

11. Write a Python program to write a list to a file.


```python
my_list = ["Apple", "Banana", "Cherry", "Mango"]

with open("output.txt", "w") as f:
    for item in my_list:
        f.write(item + "\n")

f = open("output.txt")
print(f.read())
```
