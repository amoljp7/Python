1. Write a Python function to find the maximum of three numbers.

---



```python
def maximum_of_three(a,b,c):
  if a>b & a>c:
    return a
  elif b>c & b>a:
    return b
  else:
    return c

print(maximum_of_three(10,20,30))
```

2. Write a Python function to sum all the numbers in a list.

	Sample List : (8, 2, 3, 0, 7)
	Expected Output : 20

```python
def sum_of_all(list1):
  sum = 0
  for item in list1:
    sum = sum + item
  return sum

number = [8,2,3,0,7]
print(sum_of_all(number))

```

7. Write a Python function that takes a list and returns a new list with distinct elements from the first list.

	Sample List : [1,2,3,3,3,3,4,5]
	Unique List : [1, 2, 3, 4, 5]


```python
def unique_list_maker(list1):
  ans = []
  for item in list1:
    if item not in ans:
      ans.append(item)
  return ans

l1 = [1,2,3,3,3,3,4,5]
print(unique_list_maker(l1))

```

8. Write a Python function that checks whether a passed string is a palindrome or not.

	Note: A palindrome is a word, phrase, or sequence that reads the same backward as forward, e.g., madam or nurses run.

```python
def check_pal():
  str = input("enter the string")
  revstr = ''
  for char in str:
    revstr = char + revstr

  if str == revstr:
    print("pallindrome")
  else:
    print("not pallindrome")

check_pal()
```

9. Write a Python program that accepts a hyphen-separated sequence of words as input and prints the words in a hyphen-separated sequence after sorting them alphabetically.

	Sample Items : green-red-yellow-black-white
	Expected Result : black-green-red-white-yellow

```python
def sort_words(s):
    words = s.split('-')
    words.sort()
    return '-'.join(words)


text = "green-red-yellow-black-white"
print(sort_words(text))
```

12. Write a Python program to create a function that takes one argument, and that argument will be multiplied with an unknown given number.

	Sample Output:
	Double the number of 15 = 30
	Triple the number of 15 = 45
	Quadruple the number of 15 = 60
	Quintuple the number 15 = 75

```python
def multiplier(n):
    def inner(x):
        return x * n
    return inner


double = multiplier(2)
triple = multiplier(3)
quadruple = multiplier(4)
quintuple = multiplier(5)


num = 15

print("Double the number of 15 =", double(num))
print("Triple the number of 15 =", triple(num))
print("Quadruple the number of 15 =", quadruple(num))
print("Quintuple the number of 15 =", quintuple(num))
```

13. Write a Python program to sort a list of tuples using Lambda.

	Original list of tuples:
	[('English', 88), ('Science', 90), ('Maths', 97), ('Social sciences', 82)]
	Sorting the List of Tuples:
	[('Social sciences', 82), ('English', 88), ('Science', 90), ('Maths', 97)]

```python
subjects = [('English', 88), ('Science', 90), ('Maths', 97), ('Social sciences', 82)]

sorted_list = sorted(subjects, key=lambda x: x[1])


print("Original list of tuples:")
print(subjects)

print("\nSorting the List of Tuples:")
print(sorted_list)
```

19. Write a Python program to rearrange positive and negative numbers in a given array using Lambda.

	[-1, 2, -3, 5, 7, 8, 9, -10]

	[2, 5, 7, 8, 9, -10, -3, -1]

22. Write a Python program to check whether a given string contains a capital letter, a lower case letter, a number and a minimum length using lambda.


```python

text = input("Enter a string: ")


has_upper = lambda s: any(c.isupper() for c in s)
has_lower = lambda s: any(c.islower() for c in s)
has_digit = lambda s: any(c.isdigit() for c in s)
min_length = lambda s: len(s) >= 8


if has_upper(text) and has_lower(text) and has_digit(text) and min_length(text):
    print("Valid string")
else:
    print("Invalid string")
```
